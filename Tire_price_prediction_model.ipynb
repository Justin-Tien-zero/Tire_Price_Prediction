```python
import pandas as pd
import numpy as np

df = pd.read_csv('Car_Tyres_Dataset.csv')
print(df.shape)
df.head(10)
```

```python
# split Size column
df[['Tire_Width0', 'Tire_Size']] = df['Size'].str.split(' ', n=1, expand=True)
df = df.drop('Size', axis=1)

df['Tire_Width0'].unique()
```

```python
df[['Tire_Width', 'Aspect_Rstio']] = df['Tire_Width0'].str.split('/', n=1, expand=True)
df = df.drop('Tire_Width0', axis=1)

df['Tire_Width'].unique()
```

```python
df['Aspect_Rstio'].unique()
```

```python
df['Aspect_Rstio'].fillna(value=0, inplace=True)
df['Aspect_Rstio'].unique()
```

```python
df['Tire_Size'].unique()
```

```python
# Define the split length as negative
split_length = -2

# Function to split the string
def split_string(s, length):
    # Use negative indexing to split
    return s[:length], s[length:]

# Apply the function and create new columns
df[['Const_type', 'Wheel_diam']] = df['Tire_Size'].apply(lambda x: pd.Series(split_string(x, split_length)))

# Display the updated DataFrame
df
```

```python
df['Const_type'] = df['Const_type'].str.replace(' ','')

df.Const_type.unique()
```

```python
df = df.drop('Tire_Size', axis=1)
df.info()
```

```python
# 重新索引 DataFrame
df = df.rename(columns={'Brand ': 'Brand'})

new_order = ['Brand', 'Model', 'Submodel', 'Tyre Brand', 'Serial No.', 'Type',
       'Load Index', 'Tire_Width', 'Aspect_Rstio', 'Const_type', 'Wheel_diam', 'Rating', 'Selling Price', 'Original Price']
df = df[new_order]

df.head(10)
```

```python
df.info()
```

```python
df['Selling Price'] = df['Selling Price'].str.replace(',','').astype(int)

df.info()
```

```python
df['Tire_Width'] = df['Tire_Width'].astype(int)
df['Tire_Width'].unique()
```

```python
df['Wheel_diam'] = df['Wheel_diam'].astype(int)
df['Wheel_diam'].unique()
```

```python
# Encode Nominal Variables
from sklearn.preprocessing import LabelEncoder

# 初始化LabelEncoder
le = LabelEncoder()

# Nominal Columns
col_to_le = ['Model',
      'Tyre Brand',
      'Serial No.',
      'Type',
      'Aspect_Rstio'
      ]

# 儲存編碼對照
le_dict = {}

# 擬合並轉換欄位

for col in col_to_le:
    df[col] = df[col].astype(str)
    df[col] = le.fit_transform(df[col])
    le_dict[col] = dict(zip(le.classes_, range(len(le.classes_))))



# 對每個欄位進行編碼
# for column in columns_to_encode:
#    le = LabelEncoder()
#    df[column] = le.fit_transform(df[column])
#    encoding_dict[column] = dict(zip(le.classes_, range(len(le.classes_))))

# 顯示結果

print("原始資料與編碼對照：")
for column, mapping in le_dict.items():
    print(f"{column} 對照：{mapping}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

X, y = df.iloc[: , : -1].values, df.iloc[: , -1].values
X.shape, y.shape
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
scaler = StandardScaler()

X_train_ss = scaler.fit_transform(X_train)
X_test_ss = scaler.transform(X_test)
```

```python
X_train_ss.shape, y_train.shape
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from xgboost import XGBRegressor
```

```python
from xgboost import XGBRegressor

model = XGBRegressor()
model.fit(X_train_ss, y_train)
```

```python
model.score(X_train_ss, y_train), model.score(X_test_ss, y_test)
```

```python
y_pred = model.predict(X_test_ss)
print(f'R2 = {r2_score(y_test, y_pred)*100:.2f}')
print(f'MSE = {mean_squared_error(y_test, y_pred)}')
print(f'MAE = {mean_absolute_error(y_test, y_pred)}')
```

```python
import streamlit as st
import pandas as pd

# 指定 CSV 文件的路徑
csv_file_path = 'path/to/your/cars.csv'  # 請替換成你的文件路徑

# 讀取 CSV 文件並轉換成 DataFrame
df = pd.read_csv(csv_file_path)

# 確保有 'car brand' 和 'car model' 這兩個欄位
if 'car brand' in df.columns and 'car model' in df.columns:
    # 獲取所有的 car brand
    car_brands = df['car brand'].unique()

    # 選擇 car brand
    selected_brand = st.selectbox('Select Car Brand', car_brands)

    # 根據選擇的 car brand 獲取對應的 car model
    filtered_models = df[df['car brand'] == selected_brand]['car model'].unique()

    # 選擇 car model
    selected_model = st.selectbox('Select Car Model', filtered_models)

    # 顯示選擇的結果
    st.write(f'Selected Brand: {selected_brand}')
    st.write(f'Selected Model: {selected_model}')
else:
    st.error("CSV file must contain 'car brand' and 'car model' columns.")
```
